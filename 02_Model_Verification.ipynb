Import libraries

```python
import joblib
import os
```

Load your saved model and supporting files

```python
# Saved artifact paths
model_path = "/Volumes/workspace/disease_prediction/models/final_disease_prediction_model.pkl"

encoder_path = "/Volumes/workspace/disease_prediction/models/disease_label_encoder.pkl"

feature_path = "/Volumes/workspace/disease_prediction/models/feature_columns.pkl"

# Check whether files exist
print("Model file exists:", os.path.exists(model_path))
print("Encoder file exists:", os.path.exists(encoder_path))
print("Feature file exists:", os.path.exists(feature_path))
```

Verify the loaded model dimensions

```python
rf_model = joblib.load(model_path)
encoder = joblib.load(encoder_path)
feature_columns = joblib.load(feature_path)

print("Model loaded:", rf_model is not None)
print("Encoder loaded:", encoder is not None)

print("Number of features:", len(feature_columns))
print("Model input features:", rf_model.n_features_in_)
print("Number of disease classes:", len(encoder.classes_))
```

Verify feature-column order

```python
if hasattr(rf_model, "feature_names_in_"):

    print(
        "Feature order matches:",
        list(rf_model.feature_names_in_) == feature_columns
    )

else:
    print(
        "Model does not store feature names. "
        "We need to compare against the original training notebook."
    )
```

Then test the prediction function

```python
import pandas as pd

patient = pd.DataFrame(
    0,
    index=[0],
    columns=feature_columns
)

symptoms = ["itching", "skin_rash", "fatigue"]

for symptom in symptoms:
    if symptom in patient.columns:
        patient.loc[0, symptom] = 1

prediction = rf_model.predict(patient)

disease = encoder.inverse_transform(prediction)

print("Input symptoms:", symptoms)
print("Predicted disease:", disease[0])
```
